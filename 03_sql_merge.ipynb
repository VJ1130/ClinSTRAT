```python
import sqlite3
import pandas as pd

clinical_final = pd.read_csv("brca_metabric/clinical_final.csv")
mrna_features = pd.read_csv("brca_metabric/mrna_features.csv")

conn = sqlite3.connect('clinstrat.db') #conn is a connection object that represents the database. It allows you to interact with the database, execute SQL queries, and manage transactions. In this case, it connects to a SQLite database file named 'clinstrat.db'. If the file does not exist, SQLite will create it.
clinical_final.to_sql('clinical', conn, if_exists='replace', index=False)
mrna_features.to_sql('mrna', conn, if_exists='replace', index=False)

query = """ 
SELECT c.*, m.*
FROM clinical c
JOIN mrna m ON c.PATIENT_ID = m.PATIENT_ID
WHERE c.CLAUDIN_SUBTYPE IS NOT NULL
  AND c.CLAUDIN_SUBTYPE != 'NC'
""" #why the """ ?The triple quotes `"""` in Python are used to define multi-line strings. In this case, they allow you to write the SQL query across multiple lines for better readability. You could also use single or double quotes for a single-line string, but using triple quotes makes it easier to format longer queries without needing to concatenate strings or use newline characters.
feature_matrix = pd.read_sql(query, conn)

# drop the duplicate PATIENT_ID column that comes from the join
feature_matrix = feature_matrix.loc[:, ~feature_matrix.columns.duplicated()]

print(feature_matrix.shape)
feature_matrix.to_csv("brca_metabric/feature_matrix.csv", index=False)
conn.close()
```
