```python
import pandas as pd
print("Pandas version:", pd.__version__)
```

```python
df = pd.read_csv('brca_metabric/data_clinical_patient.txt', sep='\t', comment='#')
#sep='\t' specifies that the file is tab-separated, and comment='#' indicates that lines starting with '#' should be treated as comments and ignored.
clinical_patient = df
display(clinical_patient.head(n=10))
```

```python
# the following line will display the summary information of the clinical_patient DataFrame, including the number of non-null entries, data types, and memory usage.
clinical_patient.info()
```

```python
# Subset the dataframe
#.copy() is used to create a new DataFrame that is a copy of the selected columns from the original DataFrame. This is important to avoid SettingWithCopyWarning in pandas, which can occur when trying to modify a slice of a DataFrame.
clinical_baseline = clinical_patient.copy()
display(clinical_baseline.head())
```

```python
# 1. Define the target variable (y) from the clean clinical_patient DataFrame.
y = clinical_baseline['CLAUDIN_SUBTYPE']

# 2. Define the feature matrix (X) by dropping the columns that are not needed for the model.
cols_to_drop = [
    'PATIENT_ID',
    'CLAUDIN_SUBTYPE',  # Target variable
    'THREEGENE', 'COHORT', 'HER2_SNP6',  # Target leakage
    'OS_MONTHS', 'OS_STATUS',
    'RFS_MONTHS', 'RFS_STATUS', 'VITAL_STATUS',  # Endpoints / Validation
    'SEX'  # Uninformative constant
]


# 3. Create feature matrix (X) by dropping the specified columns from the clean clinical_patient DataFrame.
X = clinical_baseline.drop(columns=cols_to_drop, inplace=False) #inplace=False ensures that the original DataFrame is not modified, and a new DataFrame is returned instead.
display(X.head(n=10))
```

```python
print("Shape of feature matrix (X):", X.shape)

print("Shape of target variable (y):", y.shape)
print("Target class counts:\n", y.value_counts())
print(y.value_counts(normalize=True))  # Display class distribution as percentages
print(y.value_counts(dropna=False))  # Display class distribution including NaN values
display(X.head(n=10))
```

```python
# Check for missing values in the feature matrix (X)
missing_values = X.isnull().sum()
print("Missing values in each column:\n", missing_values)
X.dtypes
```

```python
import pandas as pd
from sklearn.impute import KNNImputer
```

```python
# Impute missing values in continuous columns using KNN imputation
num_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
# .tolist() converts the Index object returned by select_dtypes into a regular Python list, which is often easier to work with for further processing or analysis.
knn_imputer = KNNImputer(n_neighbors=5)  # You can adjust n_neighbors based on your dataset
X[num_cols] = knn_imputer.fit_transform(X[num_cols])
print("Missing values in continuous columns after KNN imputation:")
print(X[num_cols].isnull().sum())
```

```python
# Distribution analysis of numerical columns using histograms and KDE plots 
# what are kde plots? KDE stands for Kernel Density Estimation. It is a non-parametric way to estimate the probability density function of a random variable. In simpler terms, KDE plots provide a smoothed curve that represents the distribution of data points in a dataset. Unlike histograms, which can be blocky and depend on bin sizes, KDE plots offer a continuous and smooth representation of the data distribution.
import matplotlib.pyplot as plt
import seaborn as sns

# Get numerical columns again, excluding the target variable if it's binary
numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
# Assuming 'data_clinical_basline' is the target and we might not want its distribution histogram in the same way

plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols, 1):
    plt.subplot(3, 3, i) # Adjust subplot grid as needed
    sns.histplot(X[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Boxplots: numeric features by subtype
plt.figure(figsize=(15,4))
for i, col in enumerate(['AGE_AT_DIAGNOSIS','NPI','LYMPH_NODES_EXAMINED_POSITIVE'], 1):
    plt.subplot(1,3,i)
    sns.boxplot(data=clinical_baseline, x='CLAUDIN_SUBTYPE', y=col)
    plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Countplots: categorical features vs subtype
cat_cols = ['CHEMOTHERAPY','ER_IHC','HORMONE_THERAPY',]
for col in cat_cols:
    plt.figure(figsize=(8,3))
    sns.countplot(data=clinical_baseline, x=col, hue='CLAUDIN_SUBTYPE')
    plt.xticks(rotation=45) 
    plt.title(col)
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams.update({"font.size": 11})

# -------------------------------------------------------------------------
# 1. Histogram / KDE Plot
# -------------------------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="AGE_AT_DIAGNOSIS", kde=True, bins=30, color="skyblue")
plt.title("Distribution of Age at Diagnosis")
plt.xlabel("Age at Diagnosis (Years)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# 2. Violin Plot (Continuous Feature Across Categorical Groups)
# -------------------------------------------------------------------------
plt.figure(figsize=(9, 5))
sns.violinplot(
    data=df,
    x="CLAUDIN_SUBTYPE",
    y="NPI",
    hue="CLAUDIN_SUBTYPE",
    palette="muted",
    inner="box",
    cut=0,
    legend=False,
)
plt.title("NPI Score Distribution Across Claudin Subtypes")
plt.xlabel("Claudin Subtype")
plt.ylabel("Nottingham Prognostic Index (NPI)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# 3. Count Plot (Categorical Distribution)
# -------------------------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.countplot(
    data=df,
    x="CLAUDIN_SUBTYPE",
    hue="CLAUDIN_SUBTYPE",
    order=df["CLAUDIN_SUBTYPE"].value_counts().index,
    palette="Blues_r",
    legend=False,
)
plt.title("Patient Count by Claudin Subtype")
plt.xlabel("Claudin Subtype")
plt.ylabel("Patient Count")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# 4. Scatter Plot (Relationship Between Two Continuous Variables)
# -------------------------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="AGE_AT_DIAGNOSIS",
    y="NPI",
    hue="VITAL_STATUS",
    alpha=0.7,
    palette="Set1",
)
plt.title("NPI vs. Age at Diagnosis")
plt.xlabel("Age at Diagnosis")
plt.ylabel("NPI Score")
plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# 5. Correlation Heatmap
# -------------------------------------------------------------------------
plt.figure(figsize=(10, 8))
num_cols = df.select_dtypes(include=[np.number]).columns
corr_matrix = df[num_cols].corr(method="pearson")

mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmax=1,
    vmin=-1,
    linewidths=0.5,
)
plt.title("Pearson Correlation Matrix (Raw Numeric Columns)")
plt.tight_layout()
plt.show()
```

```python
#Saving the cleaned feature matrix (X) and target variable (y) to CSV files for future use.
clinical_final = X.copy()
clinical_final['PATIENT_ID'] = clinical_baseline['PATIENT_ID'].values
clinical_final['CLAUDIN_SUBTYPE'] = y.values #what it does is it adds the target variable (CLAUDIN_SUBTYPE) to the feature matrix (X) so that we have a complete dataset with both features and labels for future analysis or modeling.

#sanity checks before saving the final dataset to CSV files. These checks help ensure that the data is in the expected format and that there are no issues before persisting it to disk.
print("Shape of final dataset:", clinical_final.shape)
print("\nAny duplicate PATIENT_IDs?", clinical_final['PATIENT_ID'].duplicated().sum())

#This is to check for any remaining missing values in the final dataset using the isnull() method followed by sum(). It will return the count of missing values for each column in the clinical_final DataFrame.
print("\nRemaining missing values in the final dataset:\n", clinical_final.isnull().sum())

clinical_final.to_csv('brca_metabric/clinical_final.csv', index=False) #index=False ensures that the index of the DataFrame is not written to the CSV file, which is often desirable when saving datasets for future analysis or modeling.
print("Final dataset saved to 'brca_metabric/clinical_final.csv'")
```
